##### Prompt Templates
- A Prompt template is a reusable prompt with placeholders(variables)
- example :- template = """
You are a Data Engineering expert.

Answer the following question:

{question}
"""

- the variable question {question} : this is placeholder

- in LangChain, a prompt template is a reusable mechanism designed to bundle static text instructions with dynamic user inputs before sending them to a language model.
- instead of hardcoding prompts, we define placeholders using curly braces {variable_name} which are substituted at runtime.

##### Overview
- Prompt Templates in LangChain allow developers to create reusable and dynamic prompts using variables. LangChain provides PromptTemplate for text-based prompts and ChatPromptTemplate for message-based prompts used with modern chat models. Advanced features include partial variables, few-shot prompting, example selectors, and MessagesPlaceholder for chat history. Prompt Templates help separate prompt logic from application code and are widely used in chatbots, RAG systems, and AI agents. 

##### PromptTemplates vs ChatPromptTemplates

- In LangChain, PromptTemplate creates a single block of plain text for text-completion models (like base LLMs), while ChatPromptTemplate manages a list of structured, role-based messages (System, Human, AI) designed specifically for conversational chat models (like GPT or Claude)

- ### PromptTemplate
- **Format** : Single block of text (string) with dynamic variables enclosed in curly braces.
- **Best For** : Non-chat models, text generation, summarization, or simple question-answering pipelines.
- code example :

```
from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate.from_template("Write a tagline for a {company} in {industry}.")
prompt.invoke({"company": "Acme", "industry": "tech"})
```

- ### ChatPromptTemplate
- **Format** : A lsit of messages. Each message explicitly defines a role (e.g., 'system', 'user', 'assistant') and a message body.
- **Best for** : Chatbots, multi-turn conversations, and scenarios requiring precise behavioral instructions.
- **Key Feature** : Supports message Placeholders  to dynamically inject entire conversation histories directly into the prompt.
- code example :

```
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert in {industry}."),
    ("human", "Write a tagline for {company}.")
])
prompt.invoke({"industry": "tech", "company": "Acme"})
```

````python
# with prompt Template
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    """
Your a data engineering expert.
Answer the following:
{question}
    """
)

# then 
formatted_prompt = prompt.format(
    question = "what is Spark"
)
print(formatted_prompt)
````

##### Why use of Prompt Templates?
- reusability: create once and use many times.
- maintainability: change prompt in one place.
- cleaner code: No messy f-strings every where
- Dynamic Inputs : variables insterted automatically.

##### Multiple Variables

````python
# we can have many placeholders 

prompt = PromptTemplate.from_template(
    """
    Your a role {role}.
    explain the following {topic} to a begginer.
    """
) 

print(prompt.format(
    role = "Data Engineer",
    topic = "Apache Spark"
))
````

#### How Prompt template works with LLM

````python
# normally 

response = llm.invoke("what is Spark")

# With Prompt Template 
formatted_prompt = prompt.format(
    topic = "Apachi Spark",
)

response = llm.invoke(formatted_prompt)
print(response.content)
````

````python
from openai.types.responses import response_error
# # Better way (prompt+Model)
# # langchain allows chaining :
chain = prompt | llm
# think : prompt ---> Chatmodel ---> response

response = chain.invoke(
    {
        "topic": "Apache Spark"
    }
)
print(response)
print(response.content)
print(response.response_metadata)
# this is called (langchain expression language).
````

```
State (LangGraph State)
State is the shared, persistent memory of the entire graph. It's passed between nodes automatically by LangGraph, survives across node executions, and is what makes multi-step reasoning possible.

from langgraph.graph import StateGraph

    messages: Annotated[list, operator.add]  # accumulates across nodes
    current_step: str
    retry_count: int
```

##### State

langgraph create one object :
```
State

{

question

destination

budget

flight

ticket

payment

history

status

}

```
- so here every node recivies the only one thing that is starting with the state.

```
State
↓
Node A
↓
Updated State
↓
Node B
↓
Updated State
↓
Node C
```
- so above we can see after every step(node) we are getting the updated state after every node.
- Real life analogy : Hospital patient data management (creating the file for patient and it will transfer to the every doctor, Google docs, where everyone will get edited and get the updated the document)

#### State is a Shared Notebook
- Think state is as a Notebook placed in the centre of a table.
```            
             Notebook

         (Shared State)

        /      |      \

     Node A  Node B  Node C
```
- if node A writes destination = Delhi, node B reads it and Node B writes flight = indigo after thi Node C read this. No confusion.

| Variables                        | State                                  |
| -------------------------------- | -------------------------------------- |
| Store one value at a time        | Stores the entire workflow information |
| Independent                      | Shared by all nodes                    |
| Passed as function arguments     | Passed as one object                   |
| Hard to manage in large systems  | Easy to manage                         |
| Difficult to track updates       | Updates happen in one place            |
| Best for small programs          | Best for AI workflows                  |
| No shared context                | Shared context                         |
| Doesn't naturally support graphs | Designed for graph execution           |

##### Why LangGraph Chose State
Langgraph was designed for Ai systems that can:
* Think
* Remember
* Make Decisions
* Call Tools
* USe Multiple Agents
* PAuse Execution
* Resume Later
* work in parallel

All of these require a shared source of truth. that shared source is the State.

- Without State, every node would have to manually exchange information with every other node. As workflows grow, this quickly becomes complex and error-prone.

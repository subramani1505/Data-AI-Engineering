#### Shared State
- Shared State is a single source of truth that is accessible to every node in a langgraph workflow.
- shared state example : Google Docs
``` markdown
                Google Doc

          /        |        \

     Person A  Person B  Person C

          \        |        /

             Same Document
```

#### Shared State Flow.

``` Markdown
               Shared State

                    |

        -------------------------

        |          |           |

      Node A     Node B      Node C

        |          |           |

        -------------------------

                    |

            Updated State
```

````python
# simple example

state = {
    "question": "What is LangGraph?",
    "documents": [],
    "answer": ""
}
````

````python
# node 1 recives
state
state["question"] # reads
# searches
{

    "documents": [

        "Official Documentation",

        "Tutorial"

    ]

} # after this Langgraph updated shared state
````

````python
# now node 2 receives exactly the same State
# Node 2 does Not recieve 

{

    "question": "What is LangGraph?"
}

# it recieves 
{

    "question": "What is LangGraph?",

    "documents": [

        "Official Documentation",

        "Tutorial"

    ],

    "answer": ""

}

# becuse it is udpated shared state

# now node2 reads documents and generates the answer
# and returns 
{

    "answer": "LangGraph is..."
}

# now the state becomes
{

    "question": "What is LangGraph?",

    "documents": [

        "Official Documentation",

        "Tutorial"

    ],

    "answer": "LangGraph is..."
}

# so next node 3 recives this udpated State
````

##### Excecution Flow
``` Markdown
Initial State
↓
Node A
↓
Shared State Updated
↓
Node B
↓
Shared State Updated
↓
Node C
↓
END
```

### Note : Here we have only one state it simply evolves.

Key Takeaways

* Shared State is the single source of truth for a graph execution.
* Every node works with the current Shared State.
* Nodes communicate indirectly through the Shared State, not directly with each other.
* Each node updates only the fields it is responsible for.
* Shared State is the foundation for advanced LangGraph capabilities like multi-agent workflows,
checkpointing, reducers, and parallel execution

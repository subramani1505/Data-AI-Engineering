TypedDict

#### Understand the Problem.
- create a dictionary
``` python
student = {
    "name": "Subbu",
    "age": 25,
    "department": "AI"
}
```
- Now we can access the values like below 
``` python
print(student["name"])
print(student["age"])
```

- so here we will get the output 25 , clean and no issue 
- for example another developer comes and writes the another dictinary like :
``` python
student = {
    "nam": "Subbu",
    "age": 25
}
```

- here if we observe the variable ```nam``` here no error and if we access the ```print(student[name])``` this time we wil get the ```key error: name``` so this error appears during excecution (runtime)

- For a large AI application with dozens of nodes, these mistakes become difficult to debug.

#### The need for a better structure : 
- This dictionary must contain certain fields, and those fields must have specific data types.

for example:

``` python
name --> string
age --> integer
department --> string
``` 
This is exactly what  ```typedict`` provides.


#### What is TypedDict?

- TypedDict is a special Python feature that allows you to define the expected structure and data types of a dictionary.
- consider this is a blueprint of dictionary , it does not create the dictionary , it describes what the dictionary should look like.

````python
### Creating a Typedict

from typing import TypedDict
````

````python
from typing import TypedDict

class Student(TypedDict):
    name: str
    age : int
    department : str
````

````python
# creating a actual dictionary
student = Student(
    name = "subbu",
    age = 25,
    department = "AI"
)

print(student)
````

````python
# acces the data
print(student["name"])
print(student.get("name"))
print(type(student))
````

````python
# another example 
from typing import TypedDict

class Book(TypedDict):
    title: str
    author: str
    price: float
    available: bool

# create an object
book = Book(
    title="Python Basics",
    author="John",
    price=499.99,
    available=True
)
print(book)
````

````python
# Conenct the above to langgrpah

state = {}

# instead of using state we will define the schema
from typing import TypedDict

class State(TypedDict):
    question : str
    documents : list
    answer : str
````

````python
# execution
state = State(
    question="What is LangGraph?",
    documents=[],
    answer=""
)

print(state)
````

````python
# updates
state["documents"] = [
    "LangGraph Documentation",
    "LangChain Docs"
]
````

````python
print(state)
````

````python
# now node 2 : LLm
# reads
state["documents"]
# generate answers and and updates 
state["answer"] = "LangGraph is..."

# Now state becomes
{
    "question": "What is LangGraph?",
    "documents": [
        "LangGraph Documentation",
        "LangChain Docs"
    ],
    "answer": "LangGraph is..."
}
````

````python
# Real Example: 
from typing import TypedDict

class EmailState(TypedDict):
    email_subject: str
    email_body: str
    sender_email: str
    category: str
    confidence: float
    knowledge_match: str
    generated_response: str
    status: str
````

````python
# now create the inital state

state = EmailState(
    email_subject="Aircraft Availability",
    email_body="Can I book a Gulfstream G650 for tomorrow?",
    sender_email="customer@example.com",
    category="",
    confidence=0.0,
    knowledge_match="",
    generated_response="",
    status="Received"
)
````

````python
# later different nodes update different fields.
# Category Node
state["category"] = "Booking Inquiry"
state["confidence"] = 0.97 

# knoweldge base node 
state["knowledge_match"] = "Fleet Availability Document"

# LLm node
state["generated_response"] = (
    "Yes, the Gulfstream G650 is available..."
)

# email sender node 
state["status"] = "Sent"


# Note : Every node works on the same state.
print(state)
````

### Stae Update.

- A State Update is only the information that has changed after a node finishes its work.
- ```Not the whole state , only the Changed Fields```

##### Visual understanding

``` python
{
question
documents
answer
status
}

Retrive node chnages 
documents

State Update

{
    documents
}

LangGraph combines this update with the existing Shared State.
```


````python
## Example: Define a State

state = {
    "question": "What is AI?",
    "documents": [],
    "answer": ""
}
````

````python
# retrive node 
def retrieve_node(state):
    docs = [
        "AI Book",
        "Wiki Pedia"
    ]
    return {
        "documents" : docs # see here we are not update entire state only we are updating the documents
    }

print(retrieve_node(state))
````

````python
print(state)
````

````python
## Single field update

# initial state
{
    "question" : "AI",
    "Answer" : ""
}

# node returns
{
    "answer" : "Artificial intelligence "
}

# finally the langgrpah produces  the updated state 
# here only one file changed.
{
    "question" : "AI",
    "answer" : "Artificial intelligence"
}
````

````python
# 2. Multiple field updates # one node can update multiple fields but langgrpah applies both.
{
"category": "Booking Inquiry",
"confidence": 0.97
}
````

Key Takeaways

* A State Update contains only the fields that changed.
* Nodes should return updates, not recreate the entire State.
* LangGraph applies those updates to the existing Shared State.
* One node can update one field, multiple fields, or no fields at all.
* Temporary variables should remain local unless future nodes need them.
* Returning only updates is what enables LangGraph to support advanced features like State Merging, Reducers, and Concurrent State Updates.

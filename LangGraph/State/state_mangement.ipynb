#### State Management.

state management is the process of designing, organizing, updating,maintaing the shared state throught the life cycle of langgraph application.

Think of it is like a company records

The records should be:
* organized
* Easy to read
* Easy to update
* Easy to maintain

The same idea applies to langgraph state

#### Principles for State Management

1.Store only what future nodes need.

Example :
``` python
def search_node(state):
    question = state["question"]
    query = question + " tutorial"
    results = search(query)
    return {
        "documents": results # here we are only returing the documents the future node needs it
    }
```

2. Dont Store Temporary Data

bad example
``` python
{
    "temp_query": "...",
    "counter": 5,
    "loop_index": 2,
    "last_iteration": 3
}
```

these above values only needed only inside the one node.
they dont need to belong shared state.

* Good Example :
``` python
query = "AI tutorial"
counter = 5
for i in range(10):
    ...
```

3. Group Related information
``` python 
# bad design
{
    "user_name": "...",
    "user_age": 25,
    "user_country": "India",
    "user_city": "Bangalore"
}

# good design
{
    "user": {
        "name": "Subbu",
        "age": 25,
        "country": "India",
        "city": "Bangalore"
    }
}
```

4. Use Meaningful names
``` python
# Bad example
{
    "d": [],
    "x": "",
    "a": ""
}

## maintain the good names 
{
    "documents": [],
    "summary": "",
    "answer": ""
}
```

5. Keep the state Small
``` python
{
    "pdf": 500 MB
}

# good to store like below
{
    "pdf": 500 MB
}
```

6. State should describe the workflow

``` python
# Example : looking at this state we immediately understand the workflow
{
    "question": "...",
    "documents": [...],
    "summary": "...",
    "status": "Searching"
}
```

7. Each Node Owns Specific fields
* suppose we have 3 nodes 
    1. retrieve node : {documents}
    2. LLm Node : {summary,answer}
    3. Email Node : {status}



8. Avoid Duplicate Information
9. Design for future growth
10. State is shared, Not personnel.

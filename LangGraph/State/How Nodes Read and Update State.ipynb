````python
from typing import TypedDict

class State(TypedDict):
    question: str
    documents: list
    answer: str


def retrieve_node(state: State):

    # Read from state
    question = state["question"]

    print(f"Searching for: {question}")

    # Process
    documents = [

        "LangGraph Official Documentation",

        "LangGraph Beginner Guide"

    ]

    # Update state
    return {

        "documents": documents

    }
````

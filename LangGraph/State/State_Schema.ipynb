    * What information is expected 
    * and How every Employees information should be organized.
This blank form is called a Schema. the actual values are filled later. The values chnages from employee to employee. But the ```structure(the schema)``` always remains the same.

#### What is exactly the state Schema?
- A state schema is the formal definition of what information the State can contain.
it answers questions like 
* what fields exists
* what is the name of each fields
* what type of data does each field store 
* which nodes are allowed to update the fields 
without scheama, langgraph would have no idea what the state should like like.

#### State vs State Schema 

- State Schema : think of it as an design, no actual values, only the structure.
- Actual State : During the excecution, the graph creates a real State.

### State Schema
``` python
state 

question : str
documents : list
summary : str
status : str
```

### State
``` python
{
    "question": "Explain LangGraph",
    "documents": [
        "Doc1",
        "Doc2"
    ],
    "summary": "LangGraph is...",
    "status": "Completed"
}
```


#### Why LangGraph Uses a State Schema
LangGraph requires a schema because it enables:
* Consistency – Every node uses the same field names.
* Readability – Developers can immediately understand what data flows through the graph.
* Type Safety – The expected data type of each field is known
* Maintainability – Large projects remain organized.
* Collaboration – Multiple developers can work on different nodes without conflicting field names.
* Predictability – Nodes know exactly what information should exist in the State.

#### State Schema vs Actual State

| State Schema             | Actual State             |
| ------------------------ | ------------------------ |
| Blueprint                | Real object              |
| Defines structure        | Stores values            |
| Created before execution | Changes during execution |
| Same for every run       | Different for each run   |
| Lists expected fields    | Contains actual data     |

```python
import os
from dotenv import load_dotenv
load_dotenv()
from google import genai

google_cloud_project_name = os.getenv("GOOGLE_CLOUD_PROJECT")
print(google_cloud_project_name)

# create the client
client = genai.Client()
print(client)

response = client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = "tell me step by step how setup the gcp account"
)
print(response.text)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    project="multi-agent-systems-502406",
    location="global"
)

result = llm.invoke("HI, tell me which model your using to give a response and act as a maths helpful assistance")
print(result)
```

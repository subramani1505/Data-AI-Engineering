# 🦜🔗 LangChain

**LangChain** is an open-source framework designed to build powerful applications powered by Large Language Models (LLMs). It acts as a cohesive toolkit and orchestrator, allowing developers to seamlessly connect LLMs to external data sources, memory, APIs, and tools to create intelligent applications like chatbots, Retrieval-Augmented Generation (RAG) systems, and autonomous AI agents.

---

### 🧩 Core Capabilities: What LangChain Connects
LangChain bridges the gap between LLMs and the external world. It allows you to connect to:

*   🤖 **LLMs & Chat Models** (e.g., OpenAI GPT, Anthropic Claude, Google Gemini)
*   💾 **Databases** (SQL, NoSQL, etc.)
*   🌐 **External APIs**
*   📄 **Document Loaders** (PDFs, Markdown, CSVs, txt, etc.)
*   🗄️ **Vector Store Databases** (Chroma, Pinecone, FAISS, etc.)
*   🧠 **Memory Systems** (Short-term & long-term conversational memory)
*   🛠️ **External Tools & Functions** (Search engines, calculators, custom Python execution)

---

### 💡 Analogy
> **Think of LangChain as a Lego kit for AI applications.** 
> Instead of writing custom integration code from scratch for every API or database, LangChain provides standardized components (Chains, Agents, Prompts, and Retrievers) to easily plug-and-play LLM-powered workflows.

# 🕸️ LangGraph

**LangGraph** is an extension of LangChain designed specifically for building **stateful, multi-actor applications** with LLMs. While core LangChain is optimized for linear, acyclic workflows (DAGs), LangGraph introduces the ability to create **cyclical graphs** (loops), which are essential for building sophisticated agentic behaviors and loops (e.g., agent-executor loops, reflection, and human-in-the-loop workflows).

---

### 🔑 Key Concepts in LangGraph

*   💾 **State**: A shared database or context (a schema) that passes through the graph. Nodes read from and write to this state.
*   🟢 **Nodes**: Individual steps or functions in your agent's workflow (e.g., an LLM call, a tool execution). Each node receives the current state and returns updates to it.
*   ↔️ **Edges**: Connections defining the flow between nodes.
    *   *Normal Edges*: Direct routing from Node A to Node B.
    *   *Conditional Edges*: Dynamic routing based on the outcome of a node (e.g., using an LLM to decide whether to call a tool or finish).

---

### ❓ Why was LangGraph created for LangChain?

1.  🔄 **Supporting Loops (Cyclical Workflows)**:
    Standard LangChain (and LCEL) is built around linear pipelines. Real-world agents require cyclic execution (e.g., `Ask LLM` -> `Call Tool` -> `Analyze Result` -> `Loop back to LLM`). LangGraph natively supports these cycles.
2.  🧠 **Robust State Management**:
    It automatically coordinates states across asynchronous steps, supporting checkpoints for **Time Travel** (rewinding, editing, and replaying agent runs) and **Human-in-the-loop** interactions (pausing execution for manual approval).
3.  🤝 **Multi-Agent Coordination**:
    It makes it easy to split a complex system into multiple distinct agents, each with their own state, prompts, and tools, collaborating as nodes in a shared graph.

---

### 💡 Analogy
> **If LangChain provides the building blocks (bricks), LangGraph is the nervous system.** 
> It coordinates complex, looping decisions and maintains the memory (state) of how those decisions evolve over time.

##### Constructor
- A constructor is a special method that runs automatically whenever an object is created.
- in python the constructor method is: ```__init__()```

````python
# example 
class Employee:
    def __init__(self):
        print("Employeee Created")

emp1 = Employee()

# as soon as object is created python automatically calls emp1.__init__()
# behind the schenes 
````

#### Why do we need it ? 

````python
# Suppose we create an employee without constructor

class Employee:
    pass

emp1 = Employee()

emp1.name = "Subbu"
emp1.salary = 50000

# here emp1 = Employee() will create an empty object and then we manually add , for every employee.
emp1.name
emp1.salary 

# for 100 or more employess it is repetative , so using the constructor
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

# create objects 
emp1 = Employee("Subbu", 50000)
emp2 = Employee("Ravi", 60000)

#now python automatically stores 
# self.name = name
# self.salary = salary      --> for every object.


# what happens internally?
# when we write 
emp1 = Employee("Subbu", 50000)

# Python secretly does:
emp1 = Employee()
emp1.__init__("Subbu", 50000) # The constructor runs automatically.
````

Understanding Self: it is the specific object we are currently working with

````python
# Consider 
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
````

````python
# when 
emp1 = Employee("Subbu", 50000)
````

````python
# python interprets as 
Employee.__init__(emp1, "Subbu", 50000)
````

````python
# so
self = emp1
name = "Subbu"
salary = 50000
````

````python
# then self
self.name = name

#becomes
emp1.name = "Subbu"
# and 
self.salary = salary
# becomes
emp1.salary = 50000
````

##### Constructor with Methods

````python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    
    def display(self):
        print(self.name, self.salary)

# usage
emp1= Employee("subbu","5000")
emp1.display()
````

##### Why we use Class?
* initalize object data automatically
* Avoid repetation code
* Ensure every object required values
* Set default values
* __init__() is a special method called a constructor in Python. It is automatically executed when an object is created and is primarily used to initialize the object's attributes with initial values.

````python
# we can have multiple methods 
class Employee:

    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(self.name,self.salary)

    def increment_salary(self):
        self.salary += 5000

# test
emp1 = Employee("Subbu",50000)
emp1.increment_salary()
print(emp1.salary)
````

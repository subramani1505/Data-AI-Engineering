```python
# calculating the total salary
employees = [
    {"name":"John","salary":50000},
    {"name":"Alice","salary":60000},
    {"name":"Bob","salary":55000}
]

total_salary = 0
for emp in employees:
    emp['salary']
    total_salary = total_salary + emp['salary']

print(total_salary)
```

```python
employees
```

```python
highest_employee  = employees[0]
highest_employee['salary']
for emp in employees:
    if emp['salary'] > highest_employee['salary']:
        highest_employee = emp

print(highest_employee)

    
```

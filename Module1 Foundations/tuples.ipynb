```python
# problem 2: Total sales 
employees = [
    ("John", 50000),
    ("Alice", 60000),
    ("Bob", 55000)
]

def total_salary(employees):
    #total = sum(salary for name, salary in employees)
    total = 0
    for name,salary in employees:
        total = total + salary
    return total

result = total_salary(employees)
print(result)
```

```python
# Tuple Unpacking
# take one row 
row = ("jhon","5000")
# normal
name = row[0]
salary = row[1]
print(name,salary)

# use the following method : Tuple Unpacking 
name, salary = row # the two variables matches values present in the tuple.
print(name,salary) # see here no. of values = no. of variables
```

```python
# find the highest paid employee
employees = [
    ("John", 50000),
    ("Alice", 60000),
    ("Bob", 55000)
]

names = []
salarys = []
for name, salary in employees:
    names.append(name)
    salarys.append(salary)
max_salary = max(salarys)
employe_index = (salarys.index(max_salary))
employ_name = names[employe_index]
print(employ_name)
print(max_salary)
print(names,salarys)
```

#### Streaming support for real-time responses

- Problem: imagine we asked a question gpt ```Explain Apache Spark in detail.``` the answer contains 1000 words. without streaming ```user sends question (waits a 10 seconds) entire answer appears.
- visual : question --> wiating..waiting...waiting..-->Full Answer. it leads to a bad user experience. 
users are expecting the quick / instant results from a model. so the explantion wants from the user is ```question --> A Ap apa apach apachespar``` response starts immediately. this is called ``streaming``

### Streaming 
Streaming is the process of recieving and displaying an LLm's output incrementally (token by token or chunk by chunk) instead of waiting for the entire response.

- Without streaming, your user stares at a blank screen for 5-10 seconds, then the entire response appears at once. With streaming, tokens appear word by word — just like ChatGPT. The perceived speed is dramatically better even though the total time is the same.

```
Without streaming:
User asks → [........10 seconds of silence........] → Full answer appears

With streaming:
User asks → "LangChain" → " is" → " a" → " framework" → " for" → ...
            (words appear as they're generated — feels instant)
```

##### The Core Concept: Generators and Async Generators
- langchain streaming is built top of on python generators. instead of returning one big response, a streamed component ``yield``s chunks one at a time.

````python
# normal function returns everything at once 
def get_answer():
    return "Langchain is a framework for buildng applications"
result = get_answer()
print(result)  # this returns everything at once after having the full information.
````

````python
# generator - yields piece by piece 
def stream_answer():
    words = ["LangChain", " is", " a", " framework", " for", " building", " LLM", " apps."]
    for word in words:
        yield word

for chunk in stream_answer():
    print(chunk,end="", flush=True)
````

Every .steam() call in langchain works exactly like this in under the hood.

Level 1 — Basic LLM Streaming

````python
from langchain_groq import ChatGroq
from langchain.messages import HumanMessage

llm = ChatGroq(model_name=os.getenv("groq_model_name"))

# without streaming - waits for full response
response = llm.invoke([HumanMessage("Explain Rag in Detail")])
print(response.content) # entire response came after 2.6 seconnds 
````

Level 2 — Streaming a Chain (LCEL)
- Every Lcel chain built with | automatically supports streaming. the stream passes through each component.

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in simple terms. Give 3 real-world examples."
)
llm    = llm
parser = StrOutputParser()

chain = prompt | llm | parser

# stream() — yields string chunks
print("Answer: ", end="")
for chunk in chain.stream({"topic": "vector embeddings"}):
    print(chunk, end="", flush=True)
print()  # newline at end

````

````python
# astream() — async version for web apps
import asyncio
import nest_asyncio
nest_asyncio.apply()

async def stream_async():
    print("Async stream: ", end="")
    async for chunk in chain.astream({"topic": "RAG"}):
        print(chunk, end="", flush=True)
    print()

asyncio.run(stream_async())
````

````python
# What Happens Inside
"""
chain.stream({"topic": "embeddings"})

prompt.stream()  → yields one ChatPromptValue (not streaming — fires once)
    ↓
llm.stream()     → yields AIMessageChunks one token at a time  ← streaming happens here
    ↓
parser.stream()  → yields string chunks (extracts .content from each AIMessageChunk)
    ↓
You get string chunks: "Vector", " embeddings", " are", " numerical", ...
"""
````

Level 3 — stream_events (the most powerful API)
- astream_events  gives us granural visibility into every event in the pipeline, not jsut the final output. but which component fired, what it recieved and what it returned 

````python
import asyncio
from distro import version
import nest_asyncio
nest_asyncio.apply()
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq

chain = (ChatPromptTemplate.from_template("explain {topic} in 3 sentenses in detail") 
                |
        ChatGroq(model_name=os.getenv("groq_model_name")) 
                | 
         StrOutputParser()
)

async def watch_all_events():
    async for event in chain.astream_events({"topic":"langchain"},version="v2"):
        kind = event["event"]
        name = event.get("name","")
        data = event.get("data",{})

        if kind == "on_chain_start":
            print(f"🟢 Chain started: {name}")
        
        elif kind == "on_chat_model_start":
            print(f"🧠 LLM started thinking...")
        
        elif kind == "on_chat_model_stream":
            token = data["chunk"].content
            if token:
                print(token, end="", flush=True)
        
        elif kind == "on_chat_model_end":
            print(f"\n✅ LLM finished")
        
        elif kind == "on_chain_end":
            print(f"🏁 Chain ended: {name}")

asyncio.run(watch_all_events())
````

Level 4 — Streaming a RAG Chain

````python
from langchain_community.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_chroma import Chroma
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# ── Setup (one-time) ─────────────────────────────────────
loader = PyPDFLoader("reports/annual_report.pdf")
docs   = loader.load()
chunks = RecursiveCharacterTextSplitter(
    chunk_size=1000, chunk_overlap=200
).split_documents(docs)

vectorstore = Chroma.from_documents(
    chunks, OpenAIEmbeddings(model="text-embedding-3-small"),
    persist_directory="./chroma_db"
)
retriever = vectorstore.as_retriever(search_kwargs={"k": 4})

# ── RAG Chain ─────────────────────────────────────────────
def format_docs(docs):
    return "\n\n".join(d.page_content for d in docs)

prompt = ChatPromptTemplate.from_template("""
Answer using only the context below. If not found, say "I don't know."

Context: {context}
Question: {question}
""")

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | ChatOpenAI(model="gpt-4o-mini", temperature=0)
    | StrOutputParser()
)

# ── Stream the answer ─────────────────────────────────────
question = "What was the revenue growth in Q4 2024?"

print(f"Q: {question}")
print("A: ", end="")

for chunk in rag_chain.stream(question):
    print(chunk, end="", flush=True)

print()  # final newline
````

Level 6 — Streaming Agent Responses
- Agents are trickier to stream because they have multiple steps: thinking → tool call → tool result → more thinking → answer.

````python
import asyncio
from langchain_core.tools import tool
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor

# Tools
@tool
def get_sales_data(region: str) -> str:
    """Get sales data for a specific region."""
    data = {"North": "₹4.5M", "South": "₹3.2M", "East": "₹2.8M", "West": "₹5.1M"}
    return f"{region} region sales: {data.get(region, 'Not found')}"

@tool
def get_growth_rate(region: str) -> str:
    """Get YoY growth rate for a region."""
    rates = {"North": "18%", "South": "12%", "East": "22%", "West": "34%"}
    return f"{region} growth rate: {rates.get(region, 'N/A')} YoY"

tools = [get_sales_data, get_growth_rate]
llm   = llm

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a data analyst. Use tools to answer accurately."),
    ("human", "{input}"),
    MessagesPlaceholder("agent_scratchpad")
])

agent    = create_tool_calling_agent(llm, tools, prompt)
executor = AgentExecutor(agent=agent, tools=tools, verbose=False)

# ── Stream agent events ───────────────────────────────────
async def stream_agent(question: str):
    print(f"\nQ: {question}")
    print("─" * 50)

    async for event in executor.astream_events(
        {"input": question},
        version="v2"
    ):
        kind = event["event"]
        name = event.get("name", "")
        data = event.get("data", {})

        # Tool is being called
        if kind == "on_tool_start":
            print(f"\n🔧 Tool: {name}")
            print(f"   Input: {data.get('input', '')}")

        # Tool returned result
        elif kind == "on_tool_end":
            output = data.get("output", "")
            print(f"   Result: {output}")

        # LLM generating final answer — stream tokens
        elif kind == "on_chat_model_stream":
            chunk = data.get("chunk")
            if chunk and hasattr(chunk, "content") and chunk.content:
                print(chunk.content, end="", flush=True)

    print("\n" + "─" * 50)

asyncio.run(stream_agent("Compare sales and growth for West and East regions"))
````

Level 7 — FastAPI Streaming Endpoint (production)

````python
# pip install fastapi uvicorn langchain-openai

from fastapi import FastAPI
from fastapi.responses import StreamingResponse
from fastapi.middleware.cors import CORSMiddleware
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from pydantic import BaseModel
import asyncio
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq

import nest_asyncio
import uvicorn
nest_asyncio.apply()

app = FastAPI()

# Allow frontend to call this API
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_methods=["*"],
    allow_headers=["*"]
)

# The chain
chain = (
    ChatPromptTemplate.from_template("Answer this clearly: {question}")
    | ChatGroq(model_name=os.getenv("groq_model_name"))
    | StrOutputParser()
)

class QuestionRequest(BaseModel):
    question: str

# ── Streaming endpoint ────────────────────────────────────
@app.post("/stream")
async def stream_answer(request: QuestionRequest):
    """SSE (Server-Sent Events) streaming endpoint."""

    async def generate():
        async for chunk in chain.astream({"question": request.question}):
            # SSE format: "data: <content>\n\n"
            yield f"data: {chunk}\n\n"
        yield "data: [DONE]\n\n"   # signal stream is complete

    return StreamingResponse(
        generate(),
        media_type="text/event-stream",
        headers={
            "Cache-Control": "no-cache",
            "X-Accel-Buffering": "no"   # disable nginx buffering
        }
    )

# ── Non-streaming endpoint (for comparison) ───────────────
@app.post("/ask")
async def ask_question(request: QuestionRequest):
    result = await chain.ainvoke({"question": request.question})
    return {"answer": result}

import threading
import uvicorn
import nest_asyncio

nest_asyncio.apply()

# Define a function to run the server
def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="info")

# Start the server in a separate background thread
server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()

# Run: uvicorn main:app --reload
# test on ui Swagger : http://127.0.0.1:8000/docs
````

Level 8 — Streaming with Memory (Conversational)

````python
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

session_store = {}
def get_history(session_id: str):
    if session_id not in session_store:
        session_store[session_id] = ChatMessageHistory()
    return session_store[session_id]

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant with memory."),
    MessagesPlaceholder("history"),
    ("human", "{question}")
])

chain = prompt | ChatOpenAI(model="gpt-4o-mini", streaming=True) | StrOutputParser()

chain_with_memory = RunnableWithMessageHistory(
    chain,
    get_history,
    input_messages_key="question",
    history_messages_key="history"
)

config = {"configurable": {"session_id": "user_subbu"}}

# Stream turn 1
print("Turn 1: ", end="")
for chunk in chain_with_memory.stream(
    {"question": "My name is Subbu and I work at Anthropic."},
    config=config
):
    print(chunk, end="", flush=True)
print()

# Stream turn 2 — agent remembers
print("Turn 2: ", end="")
for chunk in chain_with_memory.stream(
    {"question": "What do you know about me so far?"},
    config=config
):
    print(chunk, end="", flush=True)
print()
# "Your name is Subbu and you work at Anthropic."
````

#### Overview 

Streaming in LangChain allows LLM outputs and execution events to be delivered incrementally in real time rather than waiting for the entire response. LangChain supports streaming at the model, chain, retriever, tool, and agent levels through methods such as stream(), astream(), and astream_events(). Streaming improves user experience by reducing perceived latency and is commonly used in chat applications, RAG systems, and agentic workflows.

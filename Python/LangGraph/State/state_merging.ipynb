#### State Merging

until now we know:
1. A Node recives the current state.
2. A Node processes some information.
3. A Node returns a State Update.

Now what happens after the node returns ?

who combines the old state with new updates?
so here the concept of State merging will come into the pitcure. 
the concept is very important becuase every single node execution in langgraph ends with a State Merge.


#### State Merging
* State Merging the is the process of combining the existing shared state with the updates returned by a node to produce the next shared state. 

* lets imagine a simple workflow : user --> Node A --> Node B --> End.
* current state : 
``` python 
{
    "question": "What is LangGraph?",
    "documents": [],
    "answer": ""
}
```
* Node A returns 
``` python
{
    "documents" : [
        "official Docs",
        "tutorial"
    ]
}
```
* now the quetion: ```what should the new state looks like```
it should become
``` python
{
    "question": "What is LangGraph?",
    "documents": [
        "Official Docs",
        "Tutorial"
    ],
    "answer": ""
}
```

here the langgraph did this , this process is called State merging.

Key Takeaways

* Nodes return State Updates, not complete States.
* LangGraph is responsible for merging updates into the Shared State.
* Only the fields present in the update are changed.
* Untouched fields remain exactly as they were.
* State Merging happens after every node execution.
* Understanding State Merging is essential before learning Reducers and Parallel Execution.

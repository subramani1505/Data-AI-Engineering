#### What can a State Store.

1. Strings
``` python
state = {
    "question": "Explain LangGraph",
    "language": "English"
}
```
- strings are used for user questions, summaries, responses, status, names

2. Numbers 
``` python
state = {
    "token_count": 420,
    "temperature": 0.7,
    "confidence": 98
}
```
- numbers are commonly used for token usage, scores, confidence, cost, time.

3. Boolean values
``` python
state = {
    "token_count": 420,
    "temperature": 0.7,
    "confidence": 98
}
```
- used for decision making.

4. Lists
``` python
state = {

    "documents": [
        "PDF1",
        "PDF2",
        "PDF3"
    ]
}
```
- list of documents , messages, queries, histories etc.

5. Dictionaories
``` python 
state = {
    "user": {
        "name": "Subbu",
        "age": 25,
        "country": "India"
    }
}
```
- Here now the state becomes hierarchical. 
``` json
{
State
↓
User
↓
Name
Age
Country
}
```

6. Nested Objects
``` python
state = {
    "flight": {
        "airline": "Indigo",
        "price": 4500,
        "time": "10 AM"
    }
}
```
7. Chat History (very common)
``` python 
state = {

    "messages": [
        HumanMessage(...),
        AIMessage(...),
        HumanMessage(...)
    ]
}
```

8. Retrived Documents (for rag applications)

``` python
state = {
    "messages": [
        HumanMessage(...),
        AIMessage(...),
        HumanMessage(...)
    ]
}
```

9. Tool outputs
``` python
state = {
    "weather": {
        "temperature": 30,
        "condition": "Sunny"
    }
}
```

10. Images (state can even store the image paths)
``` python 
state = {
    "image_path": "/images/cat.jpg"
}
```

11. PDF's

``` python
state = {
    "pdf_path": "Annual_Report.pdf"
}
````
- later another nodes will reopens it.

12. Errors

``` python
state = {
    "error":
    "API Timeout"
}
```

- An error Handler nodes is reads it from here and resolve the query.

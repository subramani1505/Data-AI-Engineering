#### Pydantic 
- Pydantic is a Python library that validates, parses, and manages structured data at runtime using Python type annotations.

#### BAsemodel
- Every pydantic model starts with Basemodel.
``` python
from pydantic import BaseModel
```

- Base model as the parent class that provides.
    * validation
    * Parsing
    * Serialization
    * Helpful Methods
    * Error Reporting

- when we create our ownmodel , we inherit from BaseModel.

``` python
from pydantic import BaseModel
class Student(Basemodel):
    ....
```

````python
## pydantic model
from pydantic import BaseModel # internally behaves differently compared to the typedict

class Student(BaseModel):
    name : str
    age : int
````

````python
# creating an object 
student = Student(
    name ="subbu",
    age = 25
)
print(student)
print(type(student)) # it will create python object.
````

````python
# Now Test by passing with wrong data types

student = Student(
    name = 123,
    age = "25"
)
print(student)
````

````python
# acessing the values

student = Student(
    name ="subbu",
    age = 25
)
````

````python
print(student["name"])
````

````python
print(student.name)  # it behaves more like a python class
````

````python
### automatic type conversion
student = Student(
    name ="subbu",
    age = 25
)
````

````python
print(student.age) # Here Pydantic automatically converts it to data type converstion.
````

````python
# Nested Models

from pydantic import BaseModel

class Address(BaseModel):
    city: str
    country: str


class Employee(BaseModel):
    name: str
    age: int
    address: Address
````

````python
# create an object

employee = Employee(

    name="Subbu",

    age=25,

    address={

        "city":"Bangalore",

        "country":"India"

    }

)
````

````python
print(employee.address)
````

````python
# lists
from typing import List
from pydantic import BaseModel

class Student(BaseModel):
    name: str
    marks: List[int]
````

````python
student = Student(
    name="Subbu",
    marks=[90,95,98]
)
print(student)
````

````python
# optional Fields 
from typing import Optional
from pydantic import BaseModel

class User(BaseModel):
    name: str
    phone: Optional[str] = None

User(name = "subbu")
````

````python
# a pydantic model could be define:
from pydantic import BaseModel
from typing import Optional

class EmailState(BaseModel):
    email_subject: str
    email_body: str
    sender_email: str
    category: Optional[str] = None
    confidence: float = 0.0
    passenger_count: Optional[int] = None
    generated_response: str = ""
    status: str = "Received"
````

When Should You Use TypedDict?

Choose TypedDict when:

* You want maximum performance.
* Your State is relatively simple.
* Validation isn't necessary.
* You're following the most common LangGraph examples.
* Another layer in your application already validates the data.

When Should You Use Pydantic?

Choose Pydantic when:

* You need runtime validation.
* Your State contains nested objects.
* You want automatic type conversion.
* You need default values and optional fields.
* You want object-oriented access using dot notation.
* Data quality is critical.

#### Key Takeaways

By the end of this lesson, you should remember these five points:

1. TypedDict defines the expected structure of a dictionary but does not validate values at runtime.
2. Pydantic creates real Python objects that validate and manage data while the program is running.
3. LangGraph supports both because different applications have different needs.
4. Use TypedDict for lightweight, high-performance workflows.
5. Use Pydantic when data correctness, validation, and richer models are more important than a small performance cost.

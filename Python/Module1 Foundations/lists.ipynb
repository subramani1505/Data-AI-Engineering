```python
# List Slicing 
# it is used to extract the speicfic portion or sublist from the existing list
# list[start:stop:step]
# which retrives items from the the start index upto, but not including, stop index using a specified step interval

# core syntax parameters
# start: The index where the slice begins (inclusive). Defaults to 0 if omitted
# start: The index where the slice begins (inclusive). Defaults to 0 if omitted
# step: The increment value between each selected item. Defaults to 1 if omitted

numbers = [10,20,30,40,50]
```

```python
# first 3 elements 
e = numbers[:3]
print(e)

```

```python
# last 2 elemets 
e = numbers[-2:]
print(e)
```

```python
# list compression
# consise, one-line syntax, and easy to read and excecute fast 

# Example normal loop 
# printing a square of each number.
numbers = [1,2,3,4,5]
squares = []
for num in numbers:
    squares.append(num**2)

print(squares)
```

```python
# in list compression
# every list compession has 3 parts
# [ Expression for item in iterable]

# lets map the above one here 
# num **2 --> Expression
# num --> item
# numbers --> iterable

# so
squares = [num**2 for num in numbers]
print(squares)
```

```python
## Problem 2 : Find the failed documents
documents = [
    "SUCCESS",
    "FAILED",
    "SUCCESS",
    "FAILED",
    "FAILED"
]

def find_failed_documents(documents):
    """
    Here This finds the number of documents which are failed.
    """
    failed_count = 0
    for d in documents:
        if d == 'FAILED':
            failed_count =failed_count + 1
    return failed_count

result = find_failed_documents(documents)
print(result)
```

```python
# problem 4 : remove duplicate cutomer ids
customer_ids = [101,102,103,101,104,102,105]

def unique_customers(customer_ids):

    unique_ids = []
    
    for customer in customer_ids:
        if customer not in unique_ids: # if it is there then only append other wise no
            unique_ids.append(customer)
    return unique_ids

result = unique_customers(customer_ids)
print(result)
```

```python
# problem: 6 filter the successful documents. 

documents = [
    "SUCCESS",
    "FAILED",
    "SUCCESS",
    "PENDING",
    "SUCCESS"
]
```

```python
def successful_documents(documents):
    """
    This function will found the how many number of documents are success basically the count and status.
    """
    success_docs = []
    for doc in documents:
        if doc == "SUCCESS":
            success_docs.append(doc)
            count = success_docs.count(doc)
            #print(doc)
    return success_docs, count

result = successful_documents(documents)
print(result)
```

```python
# problem 7: Employee salary increase 
salaries = [50000, 60000, 45000, 70000]
```

```python
# each salary is increased by 10%
def increase_salary(salaries):
    """
    calculate the salary increment 
    """
    increase_salary = []
    for salary in salaries:
        inc_sal = salary + salary*0.1
        increase_salary.append(inc_sal)
        #print(increase_salary)
    return increase_salary

result = increase_salary(salaries)
print(result)
```

```python
# problem 10 : Highest Paid Employee 
employees = [
    ["John", 50000],
    ["Alice", 60000],
    ["Bob", 55000]
]

def find_highest_paid_employee(employees):
    highest_salary = 0
    highest_employee = None

    for name, salary in employees:
        if salary > highest_salary:
            highest_salary = salary
            highest_employee = name
    return highest_salary, highest_employee

result = find_highest_paid_employee(employees)
print(result)
```

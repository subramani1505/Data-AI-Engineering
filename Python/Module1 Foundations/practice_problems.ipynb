#### Problem 1: Schema Cleaner (Removing the Duplicates)
- Scenario: You ingested a list of user IDs from an external API, but due to a network glitch, some IDs were duplicated.

    - Task: Write a function to remove duplicates while maintaining the original order of ingestion.
    - Input: ['U101', 'U102', 'U101', 'U103', 'U102', 'U104']

- Expected Output: ['U101', 'U102', 'U103', 'U104']


```python

user_ids = ['U101', 'U102', 'U101', 'U103', 'U102', 'U104']

def remove_duplicates(user_ids:list)->list:
    """
    This function removing the duplicates from the user_ids and returns a unique_ids.
    return: unique_ids
    """
    unique_user_ids = []
    for id in user_ids:
        if id not in unique_user_ids:
            unique_user_ids.append(id)
    return unique_user_ids

result = remove_duplicates(user_ids)
print(f"unique user ids from the extranal api system: {result}")

```

Problem 2. Missing Value Imputation
- Scenario: You are processing sensor logs. Missing temperature readings are represented as None. You need to replace every None with a default value of 0.0.

- Input: [22.5, None, 24.1, None, 23.0]

- Expected Output: [22.5, 0.0, 24.1, 0.0, 23.0]

```python
logs = [22.5, None, 24.1, None, 23.0]

def missing_value_imputaion(logs):
    clean_logs = [ ]
    for log in logs:
        if log==None:

            log = 0.0
        clean_logs.append(log)
    return clean_logs

result = missing_value_imputaion(logs)
print(result)
```

Problem 3. API Payload Filtering
- Scenario: A downstream system only accepts non-empty string records. You need to filter out empty strings and whitespace-only records from your incoming data batch.

- Task: Filter out strings that are empty or just spaces.

- Input: ["Order_1", "", "Order_2", "   ", "Order_3"]

- Expected Output: ["Order_1", "Order_2", "Order_3"]

```python
orders_ids = ["Order_1", "", "Order_2", "   ", "Order_3"]

def api_pay_load_filtering(orders_ids):
    clean_ids = []
    for id in orders_ids:
        if id.strip() !="":
            clean_ids.append(id)
    print(clean_ids)
    return clean_ids

result = api_pay_load_filtering(orders_ids)
print(result)

```

Problem 4. Data Type Normalization
- Scenario: You extracted financial transactions from a legacy CSV where the amounts are strings with dollar signs. You need to convert them into floats for mathematical analysis.

- Task: Strip the $ sign and convert each element to a float.

- Input: ["$100.50", "$20.00", "$5.99"]

- Expected Output: [100.5, 20.0, 5.99]

```python

transactions = ["$100.50", "$20.00", "$5.99"]

def data_type_normalization(transactions):
    amounts = []
    for currency in transactions:
        cleaned_amount = currency.replace("$","")
        amounts.append(float(cleaned_amount))
    print(amounts)
    return amounts

result = data_type_normalization(transactions)
print(result)
        
```

Problem 5.Flattening Nested JSON/API Payloads
- Scenario: You fetched batches of logs from MongoDB, resulting in a nested list (a list of lists). You need to flatten this into a single list before loading it into a PostgreSQL staging table.

- Input: [[10, 20], [30, 40, 50], [60]]

- Expected Output: [10, 20, 30, 40, 50, 60]

```python
## Nested Loop Concept we should use 

input = [[10, 20], [30, 40, 50], [60]]

def flaten_list_data(input):
    flat_list = []
    for i in input:
        for j in i:
            flat_list.append(j)
    return flat_list

result = flaten_list_data(input)
print(result)

```

Problem 6. Batching/Chunking Data StreamsScenario: You have a massive list of records to insert into a database, but the target DB only allows a maximum of 3 records per INSERT statement to avoid memory issues.

- Task: Split a single list into a list of smaller "chunks" or batches of a specified size ($N$).

- Input: ['r1', 'r2', 'r3', 'r4', 'r5', 'r6', 'r7'] (with batch size $N = 3$)

- Expected Output: [['r1', 'r2', 'r3'], ['r4', 'r5', 'r6'], ['r7']]

```python
# slicing and for loop range of function.
flat_list = ['r1', 'r2', 'r3', 'r4', 'r5', 'r6', 'r7']
def insert_data_into_nested_list(flat_list):
    chunk_size = 3
    nested_list = []
    for i in range(0,len(flat_list),chunk_size):
        chunk = flat_list[i:i+chunk_size]
        nested_list.append(chunk)
    return nested_list

result = insert_data_into_nested_list(flat_list)
print(result)
```

Problem 7. Zipping Headers and Rows
- Scenario: You are parsing a raw file line-by-line. You've isolated the header row and a data row. You need to pair them up to inspect the mapping.

- Task: Combine a list of column names and a list of values into a list of tuples.

- Input: * headers = ["id", "name", "city"]

- row = [101, "Alice", "Bengaluru"]

- Expected Output: [("id", 101), ("name", "Alice"), ("city", "Bengaluru")]

```python
headers = ["id","name","city"]
row = [101,"Alice","Bengaluru"]

def map_the_data(headers,row):
    mapped_data = []
    columns = len(headers)
    for i in range(columns):
        column = headers[i]
        rows = row[i]
        pair = (column, rows)
        mapped_data.append(pair)
    return mapped_data

result = map_the_data(headers,row)
print(result)
```

Problem 8. Finding Schema Mismatches
- Scenario: You are validating incoming data batches against a master list of required metrics. You need to quickly identify which expected metrics are completely missing from the current batch.

- Task: Find elements that are present in the expected list but missing from the received list.

- Input: * expected = ["cpu_usage", "mem_usage", "disk_io", "network_in"]

- received = ["cpu_usage", "disk_io"]

- Expected Output: ["mem_usage", "network_in"]

```python
expected = ["cpu_usage", "mem_usage", "disk_io", "network_in"]
received = ["cpu_usage", "disk_io"]

def identify_missing_values(expected,received):
    missing = []
    for metric in expected:
        if metric not in received:
            missing.append(metric)
    print(missing)
    return missing

result = identify_missing_values(expected,received)
print(result)
```

Problem 9. Top-N Record Extraction (Leaderboard/Audit)
- Scenario: You need to extract the top 3 highest transaction amounts from a processing log batch for an automated fraud-detection alert.

- Task: Find the top 3 unique largest numbers from a list without manually writing complex sorting loops (hint: think about combining set() and built-in sorting/slicing).

- Input: [150, 3200, 450, 3200, 100, 2100, 90]

- Expected Output: [3200, 2100, 450]

```python

transactions = [150, 3200, 450, 3200, 100, 2100, 90]

# 1. Remove duplicate transaction amounts
unique_tx = set(transactions)

# 2. Sort the unique values in descending order (highest to lowest)
sorted_tx = sorted(unique_tx, reverse=True)

# 3. Take the first 3 elements using list slicing
top_3_fraud_alerts = sorted_tx[:3]

print(top_3_fraud_alerts)
# Output: [3200, 2100, 450]
```

Problem 10. Conditional Data Masking (PII Protection)
- Scenario: For data compliance (GDPR/Data Privacy), you need to mask sensitive customer identifiers in your pipeline logs. If an ID matches a certain format or if it's an odd integer internal ID, you must replace it with "MASKED".

- Task: Given a list of numerical IDs, replace any odd number with the string "MASKED", while keeping even numbers intact.

- Input: [1024, 1025, 1028, 1031]

- Expected Output: [1024, "MASKED", 1028, "MASKED"]

```python
customer_ids = [1024, 1025, 1028, 1031]
masked_ids = []
for cust_id in customer_ids:
    if cust_id % 2 != 0:
        masked_ids.append("Masked")
    else:
        masked_ids.append(cust_id)

print(masked_ids)
```
